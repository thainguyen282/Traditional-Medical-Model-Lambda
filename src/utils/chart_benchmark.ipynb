```python
import numpy as np
import matplotlib.pyplot as plt

# Example data
models = ['Model 1', 'Model 2', 'Model 3', 'Model 4']  # N models
tasks = ['Task A', 'Task B', 'Task C', 'Task D', 'Task E']  # M tasks
performance = np.random.rand(len(models), len(tasks))  # Random performance metrics

x = np.arange(len(tasks))  # the label locations
width = 0.2  # the width of the bars

fig, ax = plt.subplots(figsize=(12, 6))

for i, model in enumerate(models):
    offset = width * i
    rects = ax.bar(x + offset, performance[i], width, label=model)

ax.set_ylabel('Performance Score')
ax.set_title('Model Performance Across Different Tasks')
ax.set_xticks(x + width*(len(models)-1)/2)
ax.set_xticklabels(tasks)
ax.legend()
```
